```python
import numpy as np
import os
import sys
import matplotlib.pyplot as plt

notebook_path = os.getcwd()
parent_path = os.path.abspath(os.path.join(notebook_path, ".."))
sys.path.append(parent_path)
```

```python
print(parent_path)
```

```python
ds = ["ds1", "ds2"]

for i in ds :
    path = os.path.join(parent_path,"results", "inrc_train", f"best_{i}")
    train_sample_count = np.load(os.path.join(path, 'train_sample_count.npy'))
    mse_list = np.load(os.path.join(path, 'mse_list.npy'))
    plt.plot(train_sample_count, mse_list, label =f"dataset {i[-1]}")


plt.xlabel("train sample count")
plt.ylabel("MSE on test set")
plt.legend()
plt.grid()
plt.show()
```
