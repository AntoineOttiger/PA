```python
import numpy as np
import matplotlib.pyplot as plt

# Exemple : des points en 3D
n_points = 500
x = np.random.uniform(0, 10, n_points)
y = np.random.uniform(0, 10, n_points)
z = np.sin(x) * np.cos(y)  # Ou n'importe quelle valeur

# Option 1 : utiliser 2D histogram
heatmap, xedges, yedges = np.histogram2d(x, y, bins=100, weights=z)

# heatmap.shape => (100, 100)

# Affichage
plt.imshow(
    heatmap.T,     # Transposé pour que x et y soient bien orientés
    origin='lower',
    cmap='viridis'
)
plt.colorbar(label="Valeur moyenne de z")
plt.title("Heatmap 2D des points 3D")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors


def save_features_heatmap(feature_1, feature_2, score, f_name_1, f_name_2, score_name):
    #bins = min(len(np.unique(feature_1)), len(np.unique(feature_2)))
    bins = 5
    heatmap, xedges, yedges = np.histogram2d(
        x=feature_1, 
        y=feature_2, 
        #bins=int(np.sqrt(feature_1.shape[0])), 
        bins = bins,
        weights=score
    )
    counts, _, _ = np.histogram2d(
    x=feature_1,
    y=feature_2,
    bins=bins
)

    #No point = np.nan
    heatmap[counts == 0] = np.nan

    cmap = cm.viridis.copy()
    cmap.set_bad(color='white')  # couleur pour les NaN

    plt.imshow(heatmap.T, origin='lower', cmap=cmap)
    cbar = plt.colorbar()
    cbar.set_label(score_name)
    plt.xlabel(f_name_1)
    plt.ylabel(f_name_2)

    plt.plot()




    path = f"temp/fhm_{f_name_1}_{f_name_2}.png"
    
    

    return path

from itertools import product


feature_1 = np.array([1,3,4])
feature_2 = np.array([10,20,30])


combinations = np.array(list(product(feature_1, feature_2)))

feature_1 = [i[0] for i in combinations]
feature_2 = [i[1] for i in combinations]

score = np.ones(len(combinations))

save_features_heatmap(feature_1, feature_2, score, f_name_1="1", f_name_2="2", score_name="error")
```
